# 🎓 CampusPlacement-IQ: Student Placement & Salary Package Prediction Engine

**Project Architecture:**
1. **Exploratory Data Analysis** on 10,000 Indian college student records (MCA, B.Tech, BCA).
2. **Academic & Skill Correlations**: CGPA, Backlogs, Internships, and Coding/DSA Assessment.
3. **Dual Machine Learning Engine**:
   - **Classifier**: Random Forest Classifier for Placement Likelihood (Placed vs Unplaced).
   - **Regressor**: Random Forest Regressor for Salary Package (CTC in INR LPA).
4. **Career Readiness Tiers**: Institutional TPO segmentation into Tier-1 Dream vs Service IT vs At-Risk.
5. **Production Deployment**: Real-Time Streamlit Career Simulator & Power BI Executive Dashboard.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")

# 1. Load Data
df = pd.read_csv('../data/student_placement_raw.csv')
print(f"Total Students Loaded: {len(df):,}")
df.head()
```

```python
# 2. Degree-wise Placement Rate & Average CTC
placed_df = df[df['Placement_Status'] == 1]
print(f"Overall Placement Rate: {df['Placement_Status'].mean() * 100:.2f}%")
print(f"Average Salary Package: INR {placed_df['Salary_Package_LPA'].mean():.2f} LPA")
print(f"Highest Package Offered: INR {placed_df['Salary_Package_LPA'].max():.2f} LPA")

df.groupby('Degree_Stream').agg(
    Students=('Student_ID', 'count'),
    Placement_Rate=('Placement_Status', lambda x: f"{x.mean()*100:.1f}%"),
    Avg_CGPA=('CGPA', 'mean'),
    Avg_DSA=('Coding_DSA_Score', 'mean')
)
```

```python
# 3. Impact of Active Backlogs on Placement
df.groupby('Active_Backlogs')['Placement_Status'].agg(
    Total_Students='count',
    Placement_Rate=lambda x: f"{x.mean()*100:.1f}%"
)
```
